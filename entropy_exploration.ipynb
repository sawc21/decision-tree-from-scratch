```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data/game_success.csv")
print(df.columns)
df["Hit/Flop"].value_counts()
print(df["Hit/Flop"].value_counts())
```

```python
def entropy(data, target="Hit/Flop"):
    counts = data[target].value_counts()
    total = len(data)

    entropy_value = 0

    for count in counts:
        probability = count / total
        entropy_value -= probability * np.log2(probability)

    return entropy_value
```

```python
def show_split(data, feature, target="Hit/Flop"):

    print("=" * 70)
    print(f"SPLITTING ON: {feature}")
    print("=" * 70)

    print("\nDATA ENTERING THIS NODE:")
    print(data[["Record ID", feature, target]].to_string(index=False))

    parent_entropy = entropy(data, target)

    print("\nParent class counts:")
    print(data[target].value_counts())

    print(f"\nParent Entropy = {parent_entropy:.4f}")

    weighted_entropy = 0

    for value in data[feature].unique():

        subset = data[data[feature] == value]

        print("\n" + "-" * 70)
        print(f"BRANCH: {feature} = {value}")
        print("-" * 70)

        print(subset[["Record ID", feature, target]].to_string(index=False))

        print("\nClass counts:")
        print(subset[target].value_counts())

        branch_entropy = entropy(subset, target)
        weight = len(subset) / len(data)

        weighted_entropy += weight * branch_entropy

        print(f"\nRecords in branch = {len(subset)}")
        print(f"Weight = {len(subset)}/{len(data)} = {weight:.4f}")
        print(f"Branch Entropy = {branch_entropy:.4f}")

    information_gain = parent_entropy - weighted_entropy

    print("\n" + "=" * 70)
    print(f"Weighted Entropy = {weighted_entropy:.4f}")
    print(f"Information Gain for {feature} = {information_gain:.4f}")
    print("=" * 70)

    return information_gain
```

```python

print("ORIGINAL DATASET")
print()

print(df.to_string(index=False))

print("\nTarget counts:")
print(df["Hit/Flop"].value_counts())

print("\nStarting Entropy:", entropy(df))
```

```python
show_split(df, "Has High Graphics?")
```
